# 01 — Inventário dos dados

Este notebook inspeciona somente uma amostra limitada de cada base. Ele não altera `raw/` e não carrega integralmente os CSVs grandes.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT))

from src.data_inventory import build_inventory, save_inventory
```

```python
inventory = build_inventory(PROJECT_ROOT / 'raw', sample_rows=2_000)
inventory[['arquivo', 'formato', 'tamanho_mb', 'linhas_amostradas', 'observacao', 'erro_leitura']].fillna('')
```

```python
for item in inventory.to_dict(orient='records'):
    print(f"\n{item['arquivo']}")
    print('Colunas:', item.get('colunas', []))
    print('Ausentes na amostra:', item.get('ausentes_amostra', {}))
```

```python
csv_path, json_path = save_inventory(inventory, PROJECT_ROOT / 'processed' / 'inventory')
csv_path, json_path
```
